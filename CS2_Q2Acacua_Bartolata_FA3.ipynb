```python
def show(s, t="Scores"):
    print(f"\n--- {t} ---")
    for i, v in enumerate(s):
        print(f"Index {i}: {v}")

def main():
    s = [85.0, 92.5, 78.0, 88.0, 95.0, 82.0]
    while True:
        print("\n1. Display\n2. Search\n3. Average\n4. Update\n5. Display Updated\n6. Exit")
        c = input("Choice (1-6): ").strip()

        if c == '1': show(s)
        elif c == '2':
            t = float(input("Search score: "))
            matches = [i for i, x in enumerate(s) if x == t]
            print(f"Found at indices: {matches}" if matches else "Not found")
        elif c == '3':
            print(f"Average: {sum(s) / len(s):.2f}")
        elif c == '4':
            show(s)
            i = int(input(f"Enter index (0-{len(s)-1}): "))
            s[i] = float(input("Enter new score: "))
        elif c == '5': show(s, "Updated Scores")
        elif c == '6': break
        else: print("Invalid choice.")

if __name__ == "__main__":
    main()
```
