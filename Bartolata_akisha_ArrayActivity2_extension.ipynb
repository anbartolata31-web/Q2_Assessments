```python
def data_analyzer():
    # Initialize the array
    arr = ["Notebook", "Pen", "Calculator", "Laptop", "Book"]
    print("Original Array:", arr)

    # Update value using index
    idx = int(input("Input Index to Update (e.g., 2): ") or 2)
    val = input("Input New Value (e.g., Tablet): ") or "Tablet"
    arr[idx] = val

    # Slicing range
    slice_input = input("Input Slice Range start:end (e.g., 1:4): ") or "1:4"
    start, end = map(int, slice_input.split(":"))
    sub_arr = arr[start:end]
    print(f"Sliced Portion ({slice_input}):", sub_arr)

    # Search item in the sliced portion / array
    search = input("Input Search Item (e.g., Laptop): ") or "Laptop"
    found = search in sub_arr
    print(f"Search Result for '{search}' in sliced portion:", "Found" if found else "Not Found")

    # Display final updated array
    print("Display Updated Array/List:", arr)

if __name__ == "__main__":
    data_analyzer()
```
